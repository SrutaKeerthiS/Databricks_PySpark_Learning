```python
from pyspark.sql.types import StructField , StructType, IntegerType , StringType, DateType

schema =  StructType([
StructField("order_id",IntegerType(),False),
StructField("product_name",StringType(),True),
StructField("category",StringType(),True),
StructField("quantity",IntegerType(),True),
StructField("price",DecimalType(10,2),True),
StructField("city",StringType(),True),
StructField("order_date",DateType(),True)

])


df_sample_2=spark.createDataFrame(data=data,schema=schema)
```

### select transformation

```python
from pyspark.sql.functions import col, expr
df_sample_2.select(col("order_id"),col("product_name"),expr("category")).show()
```

## Casting a datatype

```python
df_sample_2.select(col("category"),expr(("cast(price as string)"))).printSchema()
```

```python
from pyspark.sql.functions import cast

df_sample_2.select(col("price").cast("string")).printSchema()
```

## SQL Like Expressions in Python

```python
df_sample_2.select(col("category"),col("city"),expr("cast(price as string)") , expr("cast(price as string)")).show()
```

```python

df_sample_2.selectExpr("sum(price) as price").show()
df_sample_2.selectExpr("city","cast(price as string)","quantity").show(1)
df_sample_2.select(expr("sum(price) as price")).show(1)

```

## Filtering the dataset

```python
df_sample_2.filter((col("quantity")>2) & (col("price")>4000)).selectExpr("*","quantity*price as saleprice").show()
df_sample_2.where((col("quantity")>2) & (col("price")>4000)).selectExpr("*","quantity*price as saleprice").show()
```

## Creating a new column

```python
df_sample_2.withColumn("saleprice",col("quantity")*col("price")).show(1)

df_sample_2.selectExpr("*","quantity*price as saleprice").show(1)
```

## Literals in python

```python
from pyspark.sql.functions import lit 
df_sample_2.withColumn("misc_charges",lit(25)).withColumn("net price",col("price")-col("misc_charges")).withColumn("net sales",col("quantity")*(col("price")-col("misc_charges"))).show(2)

df_sample_2.selectExpr("*","cast(25 as int) as misc_charges","price-25 as net_price","quantity * (price-25) as net_sales").show(2)
```

```python
df_sample_2.selectExpr("*","case when category='Electronics' then 5 else 2 end as discount").show()
```

## Multiple Transformations using a withColumn

```python
from pyspark.sql.functions import lit 
df_sample_2.withColumn("misc_charges",lit(25)).withColumn("net price",col("price")-col("misc_charges")).withColumn("net sales",col("quantity")*(col("price")-col("misc_charges"))).show(2)


## equivalent alternate code
columns={
    'misc_charges': lit(25),
    'net_price': col("price")-25,
    'net_sales': col('quantity')*(col('price')-25)
}
df_sample_2.withColumns(columns).show(2)
```
