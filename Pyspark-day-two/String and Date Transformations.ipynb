```python
import datetime,decimal
columns = [
"order_id",
"product_name",
"category",
"quantity",
"price",
"city",
"order_date"
]

data = [
    (1001, "Laptop", "Electronics", 2, decimal.Decimal("65000.50"), "Chennai", datetime.datetime(2026,3,12)),
    (1002, "Office Chair", "Furniture", 5, decimal.Decimal("8500.00"), "Bangalore", datetime.datetime(2026,3,12)),
    (1003, "Keyboard", "Electronics", 10, decimal.Decimal("1250.75"), "Hyderabad", datetime.datetime(2026,3,12)),
    (1004, "Water Bottle", "Accessories", 20, decimal.Decimal("450.50"), "Chennai", datetime.datetime(2026,3,12)),
    (1005, "Monitor", "Electronics", 4, decimal.Decimal("18500.00"), "Mumbai", datetime.datetime(2026,3,12)),
    (1006, "Study Table", "Furniture", 3, decimal.Decimal("12000.25"), "Pune",datetime.datetime(2026,3,12)),
    (1007, "Headphones", "Electronics", 8, decimal.Decimal("3200.50"), "Bangalore", datetime.datetime(2026,3,12)),
    (1008, "Backpack", "Accessories", 6, decimal.Decimal("2100.00"), "Chennai", datetime.datetime(2026,3,12)),
    (1009, "Mouse", "Electronics", 15, decimal.Decimal("850.75"), "Hyderabad", datetime.datetime(2026,3,12)),
    (1010, "Bookshelf", "Furniture", 2, decimal.Decimal("9500.00"), "Mumbai", datetime.datetime(2026,3,12))
]
```

```python
from pyspark.sql.types import StructField , StructType, IntegerType , StringType, DateType, DecimalType

schema =  StructType([
StructField("order_id",IntegerType(),False),
StructField("product_name",StringType(),True),
StructField("category",StringType(),True),
StructField("quantity",IntegerType(),True),
StructField("price",DecimalType(10,2),True),
StructField("city",StringType(),True),
StructField("order_date",DateType(),True)

])


df_sample_2=spark.createDataFrame(data=data,schema=schema)
```

## Case Statements 

 Case <br>
    when <condition is True> then <output 1> <br>
    when <condition is True> then <output 2> <br>
    else <default output> <br>
end <br>


```python
df_sample_2.show(1)

df_sample_2.selectExpr("*","case when category='Electronics' then 'E' when category='Furniture' then 'F' else 'A' end as category_code").show()
```

```python
from pyspark.sql.functions import when,col

df_sample_2.withColumn("category_code",when(col("category")=='Electronics','E').when(col("category")=='Furniture','F').otherwise('A')).show()
```

## regex_replace
creating product codes from product names by removing spaces and vowel characters with a empty string

```python
from pyspark.sql.functions import regexp_replace
df_sample_2.withColumn("product_code",regexp_replace(col("product_name"),'[aeiou ]','')).show()
```
